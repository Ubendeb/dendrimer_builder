```python
from src import DataSaver

"""
Test for Dendrimer Reaction Builder with Library Integration
"""
print("🧪 Testing Dendrimer Reaction Builder Interface with Library...")

# Initialize required components
from src.core.fragment_manager import FragmentManager
from src.interfaces.dendrimer_reactions_interface import DendrimerReactionBuilder
from src.interfaces.fragment_builder import DendrimerFragmentBuilder
from src.interfaces.fragment_data_manager import FragmentDataManager
from src.core.chemistry_operations import ChemistryOperations

# Create fragment manager and load from library
fragment_manager = FragmentManager()
try:
    fragment_manager.load_from_file('dendrimer_fragments.json')
    print("✓ Fragments loaded from library file")
except FileNotFoundError:
    print("⚠ Library file not found, creating test fragments...")

    # Create test fragments if library doesn't exist
    from src.io.format_converters import FormatConverters

    # Test fragments data
    methyl_acrylate_json = {
        "name": "Methyl_acrylate",
        "smiles": "C=CC(OC)=O",
        "connection_atoms": [0, 2],
        "replacement_groups": ["=1", "OC"],
        "num_atoms": 6
    }

    ethylenediamine_json = {
        "name": "Ethylenediamine",
        "smiles": "NCCN",
        "connection_atoms": [0, 3],
        "replacement_groups": ["H", "H"],
        "num_atoms": 4
    }

    # Convert and add to manager
    methyl_acrylate_mol = FormatConverters.json_to_metadata_mol(methyl_acrylate_json)
    ethylenediamine_mol = FormatConverters.json_to_metadata_mol(ethylenediamine_json)

    fragment_manager.add_fragment(
        fragment=methyl_acrylate_mol,
        name="Methyl_acrylate",
        connection_atoms=[0, 2],
        replacement_groups=["=1", "OC"]
    )

    fragment_manager.add_fragment(
        fragment=ethylenediamine_mol,
        name="Ethylenediamine",
        connection_atoms=[0, 3],
        replacement_groups=["H", "H"]
    )

    # Save the test library
    DataSaver().save_fragment_library(fragment_manager,'dendrimer_fragments.json')
    print("✓ Test fragments created and saved to library")

# Initialize other components
molecule_renderer = MoleculeRenderer()
dendrimer_fragment_builder = DendrimerFragmentBuilder()
chemistry_ops = ChemistryOperations()
fragment_data_manager = FragmentDataManager(dendrimer_fragment_builder, fragment_manager)

# Create reaction builder
reaction_builder = DendrimerReactionBuilder(fragment_manager, chemistry_ops, molecule_renderer)
reaction_builder.update_fragment_lists()

print("✓ Reaction builder initialized")
print(f"📚 Available fragments in library: {list(fragment_manager.fragments.keys())}")

# Display interfaces
dendrimer_fragment_builder.display_interface()
fragment_data_manager.display_interface()

reaction_builder.display_interface()

print("\n🧪 Test Instructions:")
print("1. Use 'Refresh Library' button to update fragment lists")
print("2. Select 'Ethylenediamine' as core fragment")
print("3. Add 'Methyl_acrylate' as reaction step")
print("4. Click 'Preview Reactions' to see the reaction sequence")
print("5. Click 'Execute Reactions' to generate final molecules")

print("\n✅ Test setup completed successfully!")
```
